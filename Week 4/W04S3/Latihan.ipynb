```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Code ini digunakan untuk menyiapkan library dan dataset yang akan digunakan. numpy digunakan untuk membantu pengolahan data dalam bentuk array, sedangkan pandas digunakan untuk mengolah data dalam bentuk tabel atau DataFrame.

load_iris() digunakan untuk mengambil dataset Iris yang sudah disediakan oleh Scikit-learn. Dataset ini berisi data bunga Iris yang nantinya digunakan untuk melakukan klasifikasi berdasarkan beberapa ciri atau fitur yang dimiliki.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Data dari dataset Iris kemudian diubah menjadi DataFrame menggunakan pd.DataFrame(). Nama kolom diambil dari iris.feature_names, sehingga setiap fitur memiliki nama yang sesuai. Dataset Iris memiliki empat fitur, yaitu panjang sepal, lebar sepal, panjang petal, dan lebar petal. Setelah itu ditambahkan kolom Species yang berisi jenis bunga dari setiap data.

Variabel X digunakan untuk menyimpan empat fitur yang nantinya menjadi data masukan untuk model, sedangkan y berisi species yang menjadi target prediksi. Karena Species merupakan target, kolom tersebut tidak dimasukkan ke dalam X. Dari hasil yang ditampilkan, terdapat 150 data dengan empat fitur yang digunakan sebagai input.

```python
y
```

Code ini digunakan untuk melihat isi dari target atau label yang terdapat pada variabel y. Hasilnya menunjukkan bahwa dataset Iris memiliki tiga species, yaitu setosa, versicolor, dan virginica. Ketiga nama tersebut menjadi hasil yang nantinya harus bisa dibedakan oleh model berdasarkan nilai dari empat fitur yang tersedia.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Label yang terdapat pada y masih berupa teks, sehingga perlu diubah menjadi bentuk angka agar dapat digunakan dalam proses training model. Untuk melakukan perubahan tersebut digunakan LabelEncoder dari Scikit-learn.

LabelEncoder() membuat sebuah encoder yang akan mengubah setiap nama species menjadi angka. Setelah fit_transform() dijalankan, label setosa diubah menjadi 0, versicolor menjadi 1, dan virginica menjadi 2. Dengan begitu, model tidak lagi memproses nama species dalam bentuk teks, tetapi menggunakan nilai numerik.

Hasil yang ditampilkan berupa array yang berisi angka 0, 1, dan 2. Urutan angka tersebut tetap mengikuti urutan data pada dataset sebelumnya. Nantinya, jika hasil prediksi model masih berupa angka, hasil tersebut dapat dikembalikan menjadi nama species menggunakan inverse_transform().

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Data kemudian dibagi menjadi data training dan data testing. Data training digunakan untuk melatih model, sedangkan data testing digunakan untuk melihat kemampuan model dalam melakukan prediksi terhadap data yang tidak digunakan saat proses training. Nilai test_size=0.2 berarti 20% data digunakan sebagai data testing dan 80% sisanya digunakan untuk training. Dari total 150 data, sekitar 120 data digunakan untuk training dan 30 data digunakan untuk testing. random_state=0 digunakan agar pembagian data tetap sama ketika code dijalankan kembali.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

GaussianNB digunakan untuk membuat model Naïve Bayes. Karena data pada dataset Iris berupa angka, seperti panjang dan lebar sepal serta petal, maka jenis Naïve Bayes yang digunakan adalah Gaussian Naïve Bayes.

model = GaussianNB() digunakan untuk membuat modelnya. Setelah itu, model.fit(X_train, y_train) digunakan untuk melatih model menggunakan data training yang sudah dibagi sebelumnya. X_train berisi data fitur, sedangkan y_train berisi species dari masing-masing data. Setelah proses ini selesai, model yang sudah dilatih disimpan ke dalam NB_model dan bisa digunakan untuk melakukan prediksi.

```python
result = NB_model.predict(X_test)
result
```

Code ini digunakan untuk melihat hasil prediksi dari model yang sudah dibuat sebelumnya. X_test berisi data testing yang tidak digunakan saat model dilatih. Data tersebut dimasukkan ke NB_model.predict() untuk melihat species yang diprediksi oleh model.

Hasilnya disimpan di dalam result, kemudian result ditampilkan untuk melihat seluruh hasil prediksinya. Karena sebelumnya label species sudah diubah menjadi angka menggunakan LabelEncoder, hasil yang muncul juga masih berupa angka 0, 1, dan 2. Angka tersebut masing-masing menunjukkan setosa, versicolor, dan virginica.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Code ini digunakan untuk melihat seberapa tepat hasil prediksi model pada data testing. accuracy_score() membandingkan label sebenarnya yang ada di y_test dengan hasil prediksi yang tersimpan di result.

Hasil perhitungannya kemudian dikalikan 100 supaya nilai accuracy ditampilkan dalam bentuk persen. Dari code tersebut diperoleh hasil:

Accuracy of data testing: 96.66666666666667

Jika dibulatkan, accuracy yang didapat adalah 96,67%. Artinya, dari data testing yang digunakan, sebagian besar species berhasil diprediksi dengan benar oleh model Gaussian Naïve Bayes.
